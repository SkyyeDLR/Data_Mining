```python
import numpy as np
import pandas as pd
import datetime
import matplotlib
import matplotlib.pyplot as plt
from matplotlib import colors
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
#from yellowbrick.cluster import KElbowVisualizer
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt, numpy as np
from mpl_toolkits.mplot3d import Axes3D
from sklearn.cluster import AgglomerativeClustering
from matplotlib.colors import ListedColormap
from sklearn import metrics
from sklearn.preprocessing import OrdinalEncoder
import time
import skfuzzy as fuzz
```

Display summary statistics, and state the number of samples and features in the dataset.

```python
SAVE_PATH = "pathway"

data = pd.read_csv(SAVE_PATH + "Mall_Customers.csv")

print("Shape: ", data.shape, "\n\n")

data.describe()
```

```python
data_drop_nullVals = data.copy()

if data_drop_nullVals.isnull().values.any():
  nullVals_dropped = data_drop_nullVals.shape[0] - data_drop_nullVals.dropna().shape[0]

  print("Missing values found")
  print("Values dropped: ", nullVals_dropped)
```

```python
cols_to_check = ["Age", "Annual Income (k$)", "Spending Score (1-100)"]

for col in cols_to_check:
    q1 = data[col].quantile(0.25)
    q3 = data[col].quantile(0.75)
    iqr = q3 - q1

    lower_fence = q1 - 1.5 * iqr
    upper_fence = q3 + 1.5 * iqr

    is_outlier = (data[col] < lower_fence) | (data[col] > upper_fence)

    print(col, "-> fences:", lower_fence, "to", upper_fence)
    print("   outliers found:", is_outlier.sum())
```

```python
encoder = OrdinalEncoder()
data_drop_nullVals[["Genre"]] = encoder.fit_transform(data_drop_nullVals[["Genre"]])
```

```python
cols_to_plot = ["Age", "Annual Income (k$)", "Spending Score (1-100)"]
sns.pairplot(data_drop_nullVals[cols_to_plot])
plt.show()
```

```python
#apply elbow algorithm to select the K first
features = data_drop_nullVals[["Annual Income (k$)", "Spending Score (1-100)"]]

inertia_values = []
k_values = range(1, 11)

for k in k_values:
    model = KMeans(n_clusters=k, random_state=42)
    model.fit(features)
    inertia_values.append(model.inertia_)

plt.plot(k_values, inertia_values, marker="o")
plt.xlabel("K (number of clusters)")
plt.ylabel("Inertia")
plt.show()
```

```python
K = 5

# K-Means
start = time.time()
kmeans_random = KMeans(n_clusters=K, init="random", n_init=10, random_state=42)
labels_random = kmeans_random.fit_predict(features)
end = time.time()

# K-Means++
start_pp = time.time()
kmeans_pp = KMeans(n_clusters=K, init="k-means++", n_init=10 , random_state=42)
labels_pp = kmeans_pp.fit_predict(features)
end_pp = time.time()

# Fuzzy-cmean
start = time.time()
cntr, u, u0, d, jm, p, fpc = fuzz.cluster.cmeans(
    features.T, c=K, m=2, error=0.005, maxiter=1000, seed=42
)
labels_fuzzy = np.argmax(u, axis=0)
end = time.time()
time_fuzzy = end - start

# Agglomerative Hierarchical method
start = time.time()
agg = AgglomerativeClustering(n_clusters=K, linkage="ward")
labels_agg = agg.fit_predict(features)
end = time.time()
time_agg = end - start

time_random = end - start
print("K-means (random init):", round(time_random, 4), "seconds")

time_pp = end_pp - start_pp
print("K-means (k-means++ init):", round(time_pp, 4), "seconds")

print("Fuzzy c-means:", round(time_fuzzy, 4), "seconds")

print("Agglomerative:", round(time_agg, 4), "seconds")
```
